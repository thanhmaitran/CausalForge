# DAGs, backdoor paths and adjustment sets
All numbers are computed when the notebook runs (outputs are cleared in git).

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from causalforge.graphs.dag import CausalDAG
g = CausalDAG()
for n, r in [('X','confounder'),('T','treatment'),('Y','outcome'),('M','mediator')]: g.add_variable(n, r)
for a, b in [('X','T'),('X','Y'),('T','M'),('M','Y')]: g.add_causal_edge(a, b)
print(g.backdoor_paths('T','Y'), g.adjustment_sets('T','Y'))
print('adjusting for mediator valid?', g.is_valid_adjustment_set({'M'},'T','Y'))
```

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from causalforge.synthetic.generators import make_scm
from causalforge.meta_learners.learners import *
from causalforge.evaluation.metrics import pehe
ds = make_scm(3000, 4, randomized=True, nonlinear=True)
for L in (SLearner, TLearner, XLearner):
    p = L().fit(ds.X, ds.treatment, ds.outcome).predict_cate(ds.X)
    print(L.name, round(pehe(p, ds.true_cate), 3))
```

# Cross-fitted DML and honest causal forest
All numbers are computed when the notebook runs (outputs are cleared in git).

```python
import sys; sys.path.insert(0, '../src')
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from causalforge.synthetic.generators import make_scm
from causalforge.causal_ml.dml import dml_plr
from causalforge.causal_ml.causal_forest import HonestCausalForest
ds = make_scm(3000, 5, nonlinear=True, heterogeneous=False)
r = dml_plr(ds); print('DML', r.estimate, r.ci, 'truth', ds.true_ate)
f = HonestCausalForest().fit(ds.X, ds.treatment, ds.outcome); print('forest mean CATE', f.predict_cate(ds.X).mean())
```
